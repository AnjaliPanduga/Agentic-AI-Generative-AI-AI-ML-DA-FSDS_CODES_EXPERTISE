# fibonacci sequence

```python
def fib(n):
    a=0
    b=1
    print(a)
    print(b)
    for i in range(0,n):
        c=a+b
        a=b
        b=c
        print(c)
fib(10)
```

# factorial of a number

# reduce()

```python
from functools import reduce
def add_all(a,b):
    return a+b
nums=[3,2,6,8,4,6,2]
evens=list(filter(lambda n:n%2==0,nums))
double=list(map(lambda n:n*2,evens))
sums=reduce(add_all,double)
print(evens)
print(double)
print(sums)
```

```python
nums=[3,2,6,8,4,6,2]
evens=list(filter(lambda n:n%2==0,nums))
double=list(map(lambda n:n*2,evens))
sums=reduce(lambda a,b:a+b,double)
print(evens)
print(double)
print(sums)
```

```python
nums=[3,2,6,8,4,6,2]
evens=list(filter(lambda n:n%2==0,nums))
add_2=list(map(lambda n:n+2,evens))
sub_2=list(map(lambda n:n-2,evens))
mul_2=list(map(lambda n:n*2,evens))
div_2=list(map(lambda n:n/2,evens))
adds=reduce(lambda a,b:a+b,add_2)
subs=reduce(lambda a,b:a+b,sub_2)
muls=reduce(lambda a,b:a+b,mul_2)
divs=reduce(lambda a,b:a+b,div_2)
print(evens)
print(add_2)
print(sub_2)
print(mul_2)
print(div_2)
print(adds)
print(subs)
print(muls)
print(divs)
```

```python
nums=[3,2,6,8,4,6,2]
odds=list(filter(lambda n:n%2!=0,nums))
double=list(map(lambda n:n*2,odds))
sums=reduce(lambda a,b:a+b,double)
print(odds)
print(double)
print(sums)
```
